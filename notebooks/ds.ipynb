Pandas series is a one dimensional array of indexed data. It can be created from a list or array

```python
data=pd.Series([1,2,3,4,5])
data
```

Pandas series consists sequence of values with indices, which can be accessed with values and index attributes

```python
data.values
```

```python
data.index
```

Data can also be accessed with the square brackets

```python
data[4]
```

```python
data[1:-1]
```

Index in pandas is explicitly defined hence, we can set the index as the demand

```python
data=pd.Series([1,2,3,4,5], index=['a', 'b', 'c', 'd', 'e'])
data
```

```python
data['e']
```

We can also set noncontiguous indices

```python
data=pd.Series([1,2,3,4,5], index=[3,6,8,9,11])
data
```
